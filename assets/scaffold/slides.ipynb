## Cells become slides

Each cell's **Slide Type** says where it goes:

- **Slide** starts a new slide, **Sub-Slide** a slide below it
- **Fragment** appears on the next click, like the code cell below
- **Skip** leaves a cell out, **Notes** become speaker notes

:::notes
JupyterLab sets a cell's Slide Type in the right sidebar: Common Tools, Slide Type.
